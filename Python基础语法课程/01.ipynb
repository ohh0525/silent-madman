```python
#my first python program
#我想在这个脚本中写一些代码，来实现Python变量的定义和使用。下面是一个简单的示例，展示了如何定义变量、赋值以及打印变量的值。

#定义变量
x = 10
y = 5.3
print("x =", x)
print("y =", y)

#变量的类型:数值类
print("Type of x:", type(x))
print("Type of y:", type(y))

#变量的类型:字符串类
name = "Alice"
print("name =", name)
print("Type of name:", type(name)) 

#变量的类型:布尔类
is_student = True
print("is_student =", is_student)
print("Type of is_student:", type(is_student))

#变量的类型:列表类
fruits = ["apple", "banana", "cherry"]
print("fruits =", fruits)
print("Type of fruits:", type(fruits))

#变量的类型:字典类
person = {"name": "Bob", "age": 25} 
print("person =", person)
print("Type of person:", type(person))

#变量的类型:元组类
coordinates = (10.0, 20.0)
print("coordinates =", coordinates)
print("Type of coordinates:", type(coordinates))

#变量的类型:集合类
unique_numbers = {1, 2, 3, 4, 5}
print("unique_numbers =", unique_numbers)
print("Type of unique_numbers:", type(unique_numbers)) 

#变量的类型:NoneType类
nothing = None 
print("nothing =", nothing)
print("Type of nothing:", type(nothing)) 

#变量的类型:范围类
range_values = range(1, 10)
print("range_values =", range_values)
print("Type of range_values:", type(range_values)) 
```

```python
#数值变量的运算
print("x + y =", x + y)
print("x - y =", x - y)
print("x * y =", x * y)
print("x / y =", x / y)
print("x // y =", x // y)
print("x % y =", x % y)
print("x ** y =", x ** y) 

 #字符串变量的运算
greeting = "Hello"
print("greeting =", greeting)
print("Type of greeting:", type(greeting)) 
print("greeting + name =", greeting + " " + name)
print("greeting * 3 =", greeting * 3)

#布尔变量的运算
print("is_student and True =", is_student and True)
print("is_student or False =", is_student or False)
print("not is_student =", not is_student)  

#列表变量的运算
print("fruits[0] =", fruits[0])
print("fruits[1] =", fruits[1])
print("fruits[2] =", fruits[2])

#字典变量的运算
print("person['name'] =", person['name'])
print("person['age'] =", person['age'])
```
