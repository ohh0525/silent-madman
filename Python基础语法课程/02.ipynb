```python
# Python 中的运算符

print(1 + 2)  # 加法运算
print(5 - 3)  # 减法运算
print(4 * 2)  # 乘法运算 
print(10 / 2)  # 除法运算
print(10 // 2)  # 整数除法运算
print(10 % 2)  # 取模运算
print(2 ** 3)  # 幂运算
print(5 > 3)  # 大于运算
print(5 < 3)  # 小于运算
print(5 == 3)  # 等于运算
print(5 != 3)  # 不等于运算
print(5 >= 3)  # 大于等于运算
print(5 <= 3)  # 小于等于运算
```

```python
# Python 中的逻辑运算符、成员运算符、身份运算符

# 逻辑运算符：and、or、not

print(True and False)  # and 运算
print(True or False)  # or 运算
print(not True)  # not 运算

# 成员运算符：in、not in
print(3 in [1, 2, 3])  # in 运算
print(5 not in [1, 2, 3])  # not in 运算

# 身份运算符：is、is not —— 比的是“是不是同一个对象”，不是值相不相等
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # True：值相等（==）
print(a is b)  # False：内容一样，但两个列表不是同一个对象（is）
print(a is a)  # True：就是同一个对象
print(a is not b)  # True：不是同一个对象（is not）
```

```python
# Python 中的流程控制语句

# 条件语句：if、elif、else
x = 10
if x > 0:
    print("x 是正数")
else:
    print("x 不是正数")

# 循环语句：for、while（注释要顶格，缩进的注释会被算进上一个代码块）
for i in range(5):
    print(i)  # 0 1 2 3 4

i = 0
while i < 5:
    print(i)  # 0 1 2 3 4
    i += 1  # 忘记自增就会死循环

# 跳转语句：break、continue、pass
for i in range(5):
    if i == 3:
        break  # 跳出循环
    print(i)  # 0 1 2   

for i in range(5):
    if i == 3:
        continue  # 跳过本次循环
    print(i)  # 0 1 2 4

# pass 语句
for i in range(5):
    if i == 3:
        pass  # 占位符，什么也不做
    print(i)  # 0 1 2 3 4  

# 顺序流程

print("我先执行")
print("我后执行")  # 我先执行 我后执行


```

```python
import os
import urllib.error
import urllib.parse
import urllib.request
from zai import ZhipuAiClient
from zai.core import ZaiError, APIStatusError, APITimeoutError

try:  # 可选依赖：pip install python-dotenv，没装也能正常运行
    from dotenv import load_dotenv
except ImportError:
    load_dotenv = None

# 只有 V 系列视觉模型才接受 image_url 输入；文本模型收到图片会报 1210「图片输入格式/解析错误」
VISION_MODEL = "glm-4.5v"  # 备选：glm-4v-flash（免费）、glm-4.1v-thinking-flash（支持 thinking）

def load_api_key() -> str:
    """按 环境变量 → .env 文件 的顺序读取 API Key，并做基本校验。

    Key 只保存在环境/内存中：不要 print 出来，也不要写进 notebook。
    """
    key = (os.environ.get("ZHIPUAI_API_KEY") or "").strip()
    if not key and load_dotenv is not None:
        load_dotenv()  # 读取项目根目录的 .env（记得把 .env 加进 .gitignore）
        key = (os.environ.get("ZHIPUAI_API_KEY") or "").strip()
    if not key:  # 提前失败：空 Key 会拼出非法请求头 'Bearer '，SDK 还会白重试 3 次
        raise RuntimeError(
            "未检测到 API Key：请执行 $env:ZHIPUAI_API_KEY=\"你的key\"（PowerShell），"
            "或在项目根目录的 .env 中写入 ZHIPUAI_API_KEY=你的key"
        )
    if any(ch.isspace() for ch in key):  # 含空格/换行同样会生成非法请求头
        raise ValueError("API Key 中含有空白字符，请检查配置是否被引号或换行截断。")
    return key

# 安全做法：不要把 API Key 直接写进 notebook（会被提交到 GitHub）。
# 推荐的两个来源（任选其一）：
#   1) 终端环境变量：  $env:ZHIPUAI_API_KEY="你的key"（PowerShell）/ set ...（CMD）
#   2) 项目根目录 .env：ZHIPUAI_API_KEY=你的key（需 pip install python-dotenv）
# 复用同一个 client，避免每个单元格重复建连（省去重复的 TCP/TLS 握手）；
# timeout 防止网络异常时内核被无限阻塞；max_retries 限制重试次数（默认 3 次）
# Key 用完即弃：不赋值给全局变量，避免 %whos / globals() 回显把 Key 打印出来
if "client" not in globals():
    client = ZhipuAiClient(api_key=load_api_key(), timeout=30.0, max_retries=2)

def get_reply_text(response) -> str:
    """从响应中安全取出回复文本：choices 为空或 content 为 None 都不会崩。"""
    choices = getattr(response, "choices", None)
    if not choices:  # 边界处理：choices 可能为空，直接取 [0] 会抛 IndexError
        raise RuntimeError("模型未返回任何结果（choices 为空）。")
    message = choices[0].message
    # 开启 thinking 时正文在 content，思考过程在 reasoning_content；两者都可能为 None
    text = getattr(message, "content", None) or getattr(message, "reasoning_content", None)
    return text if text else "(模型未返回文本内容，请检查 thinking 与请求参数)"

def assert_image_ok(url: str, timeout: float = 10.0) -> None:
    """发请求前自检图片链接：不可达 / 非图片 / 过大，都会变成服务端的 HTTP 400(1210)。"""
    scheme = urllib.parse.urlparse(url).scheme.lower()
    if scheme not in ("http", "https"):
        raise ValueError(f"图片链接必须是 http(s) 地址，当前是 {scheme or '空'}：{url!r}")
    for method in ("HEAD", "GET"):  # 部分图床不支持 HEAD，失败再用 GET 试一次
        try:
            req = urllib.request.Request(url, method=method, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(req, timeout=timeout) as resp:
                content_type = (resp.headers.get("Content-Type") or "").split(";")[0].strip().lower()
                if not content_type.startswith("image/"):
                    raise ValueError(
                        f"链接返回的不是图片（Content-Type={content_type!r}）："
                        "常见原因是防盗链、私有桶或链接已失效，请换成可公网直连的图片地址"
                    )
                size = int(resp.headers.get("Content-Length") or 0)
                if size > 5 * 1024 * 1024:
                    raise ValueError(f"图片 {size / 1024 / 1024:.1f}MB 超过 5MB 上限，请先压缩")
            return  # 自检通过
        except urllib.error.HTTPError as err:
            if method == "GET":  # 两种方式都拿不到，说明服务端就是不给图
                raise ValueError(f"图片链接不可访问：HTTP {err.code}（防盗链/私有桶/链接过期？）") from err

# 公网图片地址（老师原图所在的云存储已欠费失效，这里换成可公开访问的示例图：桌上两瓶啤酒）
IMAGE_URL = "https://images.pexels.com/photos/11568810/pexels-photo-11568810.jpeg?auto=compress&cs=tinysrgb&w=800"

try:
    assert_image_ok(IMAGE_URL)  # 先本地自检，别等服务端返回 400 才发现图有问题
    response = client.chat.completions.create(  # 注意：这里是 chat，不是 hat
        model=VISION_MODEL,  # 必须是视觉模型：文本模型收到图片会报 1210
        messages=[
            {
                "content": [
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": IMAGE_URL
                        }
                    },
                    {
                        "type": "text",
                        "text": "Where is the second bottle of beer from the right on the table?  Provide coordinates in [[xmin,ymin,xmax,ymax]] format"
                    }
                ],
                "role": "user"
            }
        ],
        # thinking 只有部分模型支持（如 glm-4.1v-thinking-flash），V 系列默认不开，开了反而可能报错
    )
except APITimeoutError as err:
    raise RuntimeError("请求超时（30s），请检查网络后重试。") from err
except APIStatusError as err:  # 400 / 401 / 429 / 5xx 等 HTTP 错误
    # 400 基本都是图的问题：链接不可公网访问、格式不支持，或模型不是视觉模型
    hint = "：图片无法解析，请确认链接可公网直连且用的是视觉模型" if err.status_code == 400 else ""
    raise RuntimeError(f"调用失败：HTTP {err.status_code}{hint}") from err
except ZaiError as err:
    raise RuntimeError(f"调用失败：{err}") from err

# 统一走 get_reply_text()：空 choices 抛错、空 content 回退推理内容，都有明确处理
print(get_reply_text(response))
```

```python
from zai import ZhipuAiClient
from zai.core import ZaiError, APIStatusError, APITimeoutError

missing = [n for n in ("load_api_key", "get_reply_text") if n not in globals()]
if missing:  # 单独运行本单元格时给出明确提示，避免 NameError
    raise RuntimeError(f"请先运行上一个单元格（缺少：{', '.join(missing)}）。")

# 同样从环境变量/.env 读 Key：用完即弃，不留在全局变量里
vision_model = globals().get("VISION_MODEL", "glm-4.5v")  # 复用上一个单元格的模型常量

# 复用上面已创建的 client（含 timeout 与重试设置），避免重复建连
if "client" not in globals():
    client = ZhipuAiClient(api_key=load_api_key(), timeout=30.0, max_retries=2)

image_url = globals().get("IMAGE_URL")  # 复用上一个单元格设置好的图片地址
if not image_url:  # 没跑上一格时 IMAGE_URL 不存在
    raise ValueError("未找到图片地址，请先运行上一个单元格（它会设置 IMAGE_URL）。")
if image_url.startswith(("http://", "https://")) and "assert_image_ok" in globals():
    assert_image_ok(image_url)  # 网络图片先自检，提前暴露防盗链/非图片/超大小

try:
    response = client.chat.completions.create(
        model=vision_model,
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": image_url
                        }
                    },
                    {
                        "type": "text",
                        "text": "Where is the second bottle of beer from the right on the table? Provide coordinates"
                    }
                ]
            }
        ]
    )
except APITimeoutError as err:
    raise RuntimeError("请求超时（30s），请检查网络后重试。") from err
except APIStatusError as err:
    raise RuntimeError(f"调用失败：HTTP {err.status_code}") from err
except ZaiError as err:
    raise RuntimeError(f"调用失败：{err}") from err

# 复用上一个单元格定义的 get_reply_text()：空 choices / 空 content 都有明确处理
print(get_reply_text(response))  # pyright: ignore[reportAttributeAccessIssue]
```

```python
# 函数

def greet(name: str) -> str:
    """返回一句问候语（有返回值，需要 print 才能看到结果）。"""
    return f"Hello, {name}!"

# 函数定义后必须调用才会执行
print(greet("小明"))  # Hello, 小明!
```

```python
# 类和对象

class 大学生:
    def __init__(self, 姓名, 年龄, 性别, 学校="北京大学", 专业="计算机科学与技术", 年级=1):
        # 边界校验：非法数据在创建对象时就拒绝，避免后面算出“张三正在睡-3小时”
        if not isinstance(姓名, str) or not 姓名.strip():
            raise ValueError("姓名必须是非空字符串")
        if isinstance(年龄, bool) or not isinstance(年龄, int) or 年龄 < 0:
            raise ValueError("年龄必须是非负整数")  # bool 是 int 的子类，True/False 要单独挡掉
        if 性别 not in ("男", "女", "其他"):
            raise ValueError("性别只能是 男/女/其他")
        if isinstance(年级, bool) or 年级 not in (1, 2, 3, 4):
            raise ValueError("年级只能是 1~4")
        self.name = 姓名
        self.age = 年龄
        self.gender = 性别
        self.school = 学校
        self.major = 专业
        self.grade = 年级

    def 学习(self, 课程):
        print(f"{self.name}正在学习{课程}")

    def 吃饭(self, 食物):
        print(f"{self.name}正在吃{食物}")

    def 睡觉(self, 时间):
        if isinstance(时间, bool) or not isinstance(时间, (int, float)):
            raise ValueError("睡眠时间必须是数字")  # 传字符串会在 <= 比较时抛 TypeError
        if 时间 <= 0:
            raise ValueError("睡眠时间必须大于 0")
        print(f"{self.name}正在睡{时间}小时")

    def 打游戏(self, 游戏):
        print(f"{self.name}正在玩{游戏}")

    def 参加社团(self, 社团):
        print(f"{self.name}正在参加{社团}社团活动")

    def 参加比赛(self, 比赛):
        print(f"{self.name}正在参加{比赛}比赛")

# 创建对象
学生1 = 大学生("张三", 20, "男")
学生2 = 大学生("李四", 21, "女", "清华大学", "电子信息工程", 2)

# 调用方法
学生1.学习("Python")
学生2.吃饭("面包")
学生1.睡觉(8)
学生2.打游戏("LOL")
学生1.参加社团("计算机协会")
学生2.参加比赛("ACM")

        
```
