```python
# Python 变量的定义、类型

a1 = 1

print(type(a1))  # <class 'int'>
```

```python
a2 = 1.59

print(type(a2))  # <class 'float'>
```

```python
a3 = "今天好开心"
b3 = '我今天也好开心'
c3 = """今天好开心
我今天也好开心"""
print(a3, b3, c3)  # 今天好开心 我今天也好开心 今天好开心
print(type(a3), type(b3), type(c3))  # <class 'str'> <class 'str'>  <class 'str'>
```

```python
a4 = True
b4 = False
print(type(a4), type(b4))  # <class 'bool'> <class 'bool'>
```

```python
# 循环
for i in range(5):
    print(i)  # 0 1 2 3 4
```

```python
hungry = True
if hungry:
    print("我饿了")  # 我饿了

while hungry:
    print("我饿了")  # 我饿了   
    hungry = False
    
```

```python
# Python 中的运算符

print(1 + 2)  # 加法运算
print(5 - 3)  # 减法运算
print(4 * 2)  # 乘法运算 
print(10 / 2)  # 除法运算
print(10 // 2)  # 整数除法运算
print(10 % 2)  # 取模运算
print(2 ** 3)  # 幂运算
print(5 > 3)  # 大于运算
print(5 < 3)  # 小于运算
print(5 == 3)  # 等于运算
print(5 != 3)  # 不等于运算
print(5 >= 3)  # 大于等于运算
print(5 <= 3)  # 小于等于运算
```

```python
# Python 中的逻辑运算符、成员运算符、身份运算符

# 逻辑运算符：and、or、not

print(True and False)  # and 运算
print(True or False)  # or 运算
print(not True)  # not 运算

# 成员运算符：in、not in
print(3 in [1, 2, 3])  # in 运算
print(5 not in [1, 2, 3])  # not in 运算

# 身份运算符：is、is not
print(3 is 3)  # is 运算
print(3 is not 4)  # is not 运算

```

```python
# Python 中的流程控制语句

# 条件语句：if、elif、else
x = 10
if x > 0:
    print("x 是正数")
else:
    print("x 不是正数")

    # 循环语句：for、while
for i in range(5):
    print(i)  # 0 1 2 3 4
i = 0
while i < 5:
    print(i)  # 0 1 2 3 4
    i += 1

    # 跳转语句：break、continue、pass
for i in range(5):
    if i == 3:
        break  # 跳出循环
    print(i)  # 0 1 2   

for i in range(5):
    if i == 3:
        continue  # 跳过本次循环
    print(i)  # 0 1 2 4

    # pass 语句
for i in range(5):
    if i == 3:
        pass  # 占位符，什么也不做
    print(i)  # 0 1 2 3 4  

# 顺序流程

print("我先执行")
print("我后执行")  # 我先执行 我后执行


```

```python
# 分支流程
x = 10
if x > 0:
    print("x 是正数")
else:
    print("x 不是正数")
    
```
