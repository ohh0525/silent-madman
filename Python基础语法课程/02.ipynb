```python
hungry = True
if hungry:
    print("我饿了")  # 我饿了

while hungry:
    print("我饿了")  # 我饿了   
    hungry = False
    
```

```python
# Python 中的运算符

print(1 + 2)  # 加法运算
print(5 - 3)  # 减法运算
print(4 * 2)  # 乘法运算 
print(10 / 2)  # 除法运算
print(10 // 2)  # 整数除法运算
print(10 % 2)  # 取模运算
print(2 ** 3)  # 幂运算
print(5 > 3)  # 大于运算
print(5 < 3)  # 小于运算
print(5 == 3)  # 等于运算
print(5 != 3)  # 不等于运算
print(5 >= 3)  # 大于等于运算
print(5 <= 3)  # 小于等于运算
```

```python
# Python 中的逻辑运算符、成员运算符、身份运算符

# 逻辑运算符：and、or、not

print(True and False)  # and 运算
print(True or False)  # or 运算
print(not True)  # not 运算

# 成员运算符：in、not in
print(3 in [1, 2, 3])  # in 运算
print(5 not in [1, 2, 3])  # not in 运算

# 身份运算符：is、is not —— 比的是“是不是同一个对象”，不是值相不相等
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # True：值相等（==）
print(a is b)  # False：内容一样，但两个列表不是同一个对象（is）
print(a is a)  # True：就是同一个对象
print(a is not b)  # True：不是同一个对象（is not）
```

```python
# Python 中的流程控制语句

# 条件语句：if、elif、else
x = 10
if x > 0:
    print("x 是正数")
else:
    print("x 不是正数")

    # 循环语句：for、while
for i in range(5):
    print(i)  # 0 1 2 3 4
i = 0
while i < 5:
    print(i)  # 0 1 2 3 4
    i += 1

    # 跳转语句：break、continue、pass
for i in range(5):
    if i == 3:
        break  # 跳出循环
    print(i)  # 0 1 2   

for i in range(5):
    if i == 3:
        continue  # 跳过本次循环
    print(i)  # 0 1 2 4

    # pass 语句
for i in range(5):
    if i == 3:
        pass  # 占位符，什么也不做
    print(i)  # 0 1 2 3 4  

# 顺序流程

print("我先执行")
print("我后执行")  # 我先执行 我后执行


```

```python
import os
from zai import ZhipuAiClient

# 安全做法：不要把 API Key 直接写进 notebook（会被提交到 GitHub）。
# 先在终端设置环境变量，再运行本单元格：
#   终端(PowerShell):  $env:ZHIPUAI_API_KEY="你的key"
#   终端(CMD):         set ZHIPUAI_API_KEY=你的key
client = ZhipuAiClient(api_key=os.environ.get("ZHIPUAI_API_KEY", ""))  # 从环境变量读取 API Key
response = client.chat.completions.create(
    model="glm-5.3-flash",  # 填写需要调用的模型名称
    messages=[
        {
            "content": [
                {
                    "type": "image_url",
                    "image_url": {
                        "url": "https://cloudcovert-1305175928.cos.ap-guangzhou.myqcloud.com/%E5%9B%BE%E7%89%87grounding.PNG"
                    }
                },
                {
                    "type": "text",
                    "text": "Where is the second bottle of beer from the right on the table?  Provide coordinates in [[xmin,ymin,xmax,ymax]] format"
                }
            ],
            "role": "user"
        }
    ],
    thinking={
        "type":"enabled"
    }
)
print(response.choices[0].message)
```

```python
import os
from zai import ZhipuAiClient

# 同样从环境变量读 Key，不要把 Key 写进文件
client = ZhipuAiClient(api_key=os.environ.get("ZHIPUAI_API_KEY", ""))

response = client.chat.completions.create(
    model="glm-5.3-flash",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image_url",
                    "image_url": {
                        "url": "你的图片链接"
                    }
                },
                {
                    "type": "text",
                    "text": "Where is the second bottle of beer from the right on the table? Provide coordinates"
                }
            ]
        }
    ]
)
print(response)
```
