## 0. 先修自检

列表按**编号**取，字典按**名字**取。先用两行确认字典的手感。

````python
tel = {"小明": 138, "小红": 139}
print(tel["小明"])
print(len(tel))
````

**预期输出**

```
138
2
```

- 字典存的是"**名字 → 内容**"这样**一对一对**的东西；
- 取值不用位置，直接用名字 —— 这个"名字"叫**键**。

到这里你应该已经有感觉了：列表要问"第几个"，字典要问"叫什么"。

## 1. 创建：一个键配一个值

写法是 `{键: 值, 键: 值}` —— 冒号连接一对，逗号分隔多对。

````python
person = {"name": "小明", "score": 90, "class": "一班"}
print(person)
print(person["name"])
print(person["score"])
````

**预期输出**

```
{'name': '小明', 'score': 90, 'class': '一班'}
小明
90
```

取值写 `person["name"]` —— 用**键**去取**值**。

这就是字典存在的全部理由：当你要按某个有意义的名字取数据时，用位置去记既容易错、也读不懂。

> 术语上要分清：`name` 是**键**，`小明` 是**值**。**键不是索引** —— 索引是位置编号，键是名字。

---

## 2. 分岔口：`d[键]` 会报错，`get()` 给默认值

这是本篇最重要的一条分岔口。

````python
tel = {"小明": 138, "小红": 139}
print(tel.get("小明"))      # 键存在 → 拿到 138
print(tel.get("小刚"))      # 键不存在 → 给 None（不报错）
print(tel.get("小刚", 0))   # 键不存在 → 给你指定的默认值
````

**预期输出**

```
138
None
0
```

| 写法 | 键存在 | 键不存在 |
| --- | --- | --- |
| `d[键]` | 拿到值 | **抛 `KeyError`** |
| `d.get(键)` | 拿到值 | 给 `None` |
| `d.get(键, 默认值)` | 拿到值 | 给你指定的默认值 |

规律很简单：**不确定键在不在，就用 `get()`**。这一条能消掉一整类报错。

那 `d[键]` 取不到时到底长什么样？看下面。

````python
tel = {"小明": 140, "小刚": 137}
print(tel["小明"])

# 把下面这一行行首的 # 去掉再运行，看看会发生什么：
# print(tel["小红"])
````

**预期输出**（注释保留时）

```
140
```

把最后那一行的 `#` 去掉再运行，你会看到：

```
---------------------------------------------------------------------------
KeyError                                  Traceback (most recent call last)
Cell In[4], line 5
----> 5 print(tel["小红"])

KeyError: '小红'
```

`KeyError: '小红'` 的意思是：**字典里没有 `'小红'` 这个键**。

这里要记住的是**结论**，不是读报错的技巧：想安全取值就用 `get()`，想直接取值就得先确认键存在。

> 完整的 Traceback 逐行读法与排查动作，见补充章节《报错怎么读》。本篇只负责让你认出"键不存在"这一类问题。

## 3. 增 / 改 / 删：新增和修改写法完全一样

这一点和列表很不一样，值得单独看一遍。

````python
tel = {"小明": 138, "小红": 139}

tel["小刚"] = 137      # 这个键本来不存在 → 新增
print(tel)

tel["小明"] = 140      # 这个键已存在 → 修改（写法一模一样）
print(tel)

del tel["小红"]        # 删掉整个键值对
print(tel)
````

**预期输出**

```
{'小明': 138, '小红': 139, '小刚': 137}
{'小明': 140, '小红': 139, '小刚': 137}
{'小明': 140, '小刚': 137}
```

`tel["小刚"] = 137` 和 `tel["小明"] = 140` 写法**完全相同**。

字典不在乎那个键本来有没有，它只管"现在这个键对应什么值" —— 这是它和列表在"改"上最大的不同。

> `del d[键]` 删的是**整个键值对**。想只清空内容、保留这个键，那是另一回事，本篇不展开。

## 4. 键的规矩（一）：重复会覆盖，不报错

键必须唯一 —— 但这条是**约束**，不是**报错**。

````python
scores = {"小明": 90, "小红": 85, "小明": 100}
print(scores)
print(scores["小明"])
````

**预期输出**

```
{'小明': 100, '小红': 85}
100
```

我写了**两次** `"小明"`，程序**没有报错** —— 后写的 `100` 把先写的 `90` 覆盖掉了，而且字典里最终只有**一个** `"小明"`。

对比一下：列表可以有重复元素（`[1, 1, 2]` 完全合法），字典的键不行。这就是"键必须唯一"的含义。

---

## 5. 键的规矩（二）：必须是不可变类型

字符串、数字、元组都可以当键；**列表不行**。先看三种可以的情况。

````python
# 字符串当键 —— 可以
ok1 = {"name": "小明"}
print(ok1)

# 数字当键 —— 可以
ok2 = {1: "一", 2: "二"}
print(ok2)

# 元组当键 —— 可以
ok3 = {(1, 2): "一个坐标"}
print(ok3)

# 把下面这一行行首的 # 去掉再运行，看看会发生什么：
# bad = {["name"]: "小明"}
````

**预期输出**（注释保留时）

```
{'name': '小明'}
{1: '一', 2: '二'}
{(1, 2): '一个坐标'}
```

把最后那一行的 `#` 去掉再运行，你会看到：

```
---------------------------------------------------------------------------
TypeError                                 Traceback (most recent call last)
Cell In[7], line 14
----> 14 bad = {["name"]: "小明"}

TypeError: unhashable type: 'list'
```

`TypeError: unhashable type: 'list'` 的意思是：**列表这种类型不能当键**。

原因很直白：字典靠"键这个名字"来定位内容，如果这个名字本身还能被**原地改掉**（列表就是可变的），那这本通讯录就废了。所以**键必须不可变**，才能被稳稳地定位。

上面第三种能当键的**元组** `(1, 2)`，正是因为它不可变。

> "可哈希"背后的哈希表原理超出本篇范围。你只要先建立"按键直取、不靠从头逐个找"这一层直觉就够了。

## 6. 成员运算与三个视图：`in` 只查**键**

这一节有个几乎人人都会误解一次的点。

````python
tel = {"小明": 140, "小刚": 137}

print("小明" in tel)      # 键在不在 → True
print(140 in tel)        # 140 是「值」，而 in 不查值 → False

print(list(tel.keys()))
print(list(tel.values()))
print(list(tel.items()))
````

**预期输出**

```
True
False
['小明', '小刚']
[140, 137]
[('小明', 140), ('小刚', 137)]
```

**最容易误解的一点**：`140` 明明就在字典里（是小明的电话），`in` 却说是 `False`。因为 `in` 查的是**键**那一栏，**不查值**。

| 写法 | 看什么 | 本例结果 |
| --- | --- | --- |
| `键 in d` | 键（只有键） | `True` / `False` |
| `d.keys()` | 所有**键** | `['小明', '小刚']` |
| `d.values()` | 所有**值** | `[140, 137]` |
| `d.items()` | 所有**键值对** | `[('小明', 140), ('小刚', 137)]` |

所以：判断某个**值**在不在，得看 `values()` —— 这与"键在不在"是两件事，别混。

> **成员运算 `in` 归本篇**（以及《列表》篇）；《运算符》篇只讲算术、比较、逻辑三组。

## 7. 两件小事：保持插入顺序、字典同样可变

### 7.1 字典记住你放入的先后

````python
config = {}
config["theme"] = "dark"
config["font_size"] = 16
config["show_tips"] = True
config["theme"] = "light"      # 改一个已存在的键
````

上面这段没有任何输出 —— 别急，先在心里猜一下 `config` 现在长什么样，再运行下一格。

````python
print(config)
````

**预期输出**

```
{'theme': 'light', 'font_size': 16, 'show_tips': True}
```

两个观察：

1. 打印出来是**我放入的顺序**（theme → font_size → show_tips），不是随机乱跳 —— 调试时省心；
2. `theme` 虽然最后被改过，**位置仍在最前面** —— 修改一个已存在的键**不会**把它挪到末尾。

> 较新的 Python（3.7 起）保证字典保持插入顺序。不要再把它当作"无序容器"来理解。

### 7.2 字典同样可变

字典也是**可变**的 —— 原理与《列表》篇的 `b = a` 完全一样，这里只提一句。

````python
tel = {"小明": 140, "小刚": 137}
d2 = tel              # 不是复制，是同一个字典的第二个名字
tel["新键"] = 1

print("d2 =", d2)
print("是同一个字典吗？", d2 is tel)
````

**预期输出**

```
d2 = {'小明': 140, '小刚': 137, '新键': 1}
是同一个字典吗？ True
```

和列表一模一样：`d2 = tel` 只是多了个名字，改一个两边都变。想要独立副本，用 `tel.copy()`。

> `b = a` 的完整讲解在《列表》篇；本篇只提一句"字典同样可变"，不重复展开。

## 8. 场景：给"查不到"留一个默认值

这是一个把 `get()` 用出价值的典型场景：统计某项出现了几次。

普通的写法是"取出来、加 1、放回去"，可第一次遇到某个键时它还不存在 —— 直接 `d[键] + 1` 会抛 `KeyError`。`get()` 正好补上这一步。

````python
counts = {}

counts["苹果"] = counts.get("苹果", 0) + 1    # 第一次出现：get 给 0，加 1 得 1
counts["香蕉"] = counts.get("香蕉", 0) + 1
counts["苹果"] = counts.get("苹果", 0) + 1    # 又出现一次：get 取到 1，加 1 得 2

print(counts)
````

**预期输出**

```
{'苹果': 2, '香蕉': 1}
```

把 `counts.get("苹果", 0) + 1` 拆开读：

1. `counts.get("苹果", 0)` —— 有就取出来，没有就给 `0`；
2. `+ 1` —— 加一；
3. 再赋值回 `counts["苹果"]`。

"键可能不存在"这一整类 `KeyError` 就这样从根上被消掉了。

> 真实统计会把这三行放进循环里对一大批数据跑 —— 那是**后续课程**的内容。本篇只把这一步的逻辑讲清。
