## 1.2 变量与数据类型

变量可以理解为"给一个数据贴上的标签"。Python 中不需要事先声明类型，直接赋值即可。

````python
name = "张三"
age = 20
height = 1.75
is_student = True

print(name, age, height, is_student)
````

**预期输出**

```
张三 20 1.75 True
```

### Python 的四种基本数据类型

| 类型 | 名称 | 示例 | 典型用途 |
| --- | --- | --- | --- |
| `int` | 整数 | `20`、`-3`、`1000` | 人数、件数、次数 |
| `float` | 浮点数 | `1.75`、`3.14`、`0.5` | 价格、比例、均值 |
| `str` | 字符串 | `"张三"`、`"2024-01"` | 姓名、分类、日期文本 |
| `bool` | 布尔值 | `True`、`False` | 是否通过、是否合格 |

用 `type()` 可以查看任意值的类型。**这是你排查类型问题的第一工具。**

````python
print(type(20))
print(type(1.75))
print(type("张三"))
print(type(True))
````

**预期输出**

```
<class 'int'>
<class 'float'>
<class 'str'>
<class 'bool'>
```

注意：`"张三"` 有引号，所以是 `str`；如果写成 `张三`（没有引号），Python 会以为你在引用一个叫"张三"的变量，从而报 `NameError`。

### 类型转换（本课的隐性重点）

下面这段代码演示了最容易踩的坑：**带引号的 `"20"` 是文本，不是数字。**

````python
a = "20"          # 这是字符串，不是数字
print(type(a))

b = int(a)        # 转成整数
print(b + 5)      # 可以参与运算了
````

**预期输出**

```
<class 'str'>
25
```

下面这段先注释掉第 3 行，运行看看类型；再把注释去掉，运行看结果。**动手改比看一遍有效得多。**

````python
a = "20"
print(type(a))

b = int(a)        # 试试把这行注释掉，后面的 b + 5 会怎样？
print(b + 5)
````

反过来，数字也能转成字符串：

````python
score = 95
print("我的成绩是 " + str(score) + " 分")
````

**预期输出**

```
我的成绩是 95 分
```

### ⚠️ 一个必须记住的反例

把数字字符串和数字直接相加会报错。下面是真实会出现的报错：

```python
"20" + 5
```

```
---------------------------------------------------------------------------
TypeError                                 Traceback (most recent call last)
Cell In[1], line 1
----> 1 "20" + 5

TypeError: can only concatenate str (not "int") to str
```

**这个报错怎么读？** 三个要点：

1. **最后一行** `TypeError: can only concatenate str (not "int") to str` —— 这才是结论：字符串只能和字符串相加。
2. **`----> 1`** —— 箭头指向出错的那一行，行号是 1。
3. **中间那段** —— 调用链，本格只有一行代码，所以很短；代码长了它会有好多层。

修正办法是转换类型：`"20"` 改成 `int("20")`，或者 `5` 改成 `str(5)`。

> 完整的读报错方法见补充章节《报错怎么读》。

> **划重点**：后续读取 csv 文件时，**所有读进来的内容默认都是字符串**。因此"看到数字先想想它是不是字符串"是一个必须养成的习惯——这也是第 3 课文件读写、第 4 课数据清洗反复出现的问题根源。

**参考答案**

