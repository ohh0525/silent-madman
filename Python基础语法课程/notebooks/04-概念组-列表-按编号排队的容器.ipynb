## 0. 先修自检

列表是"**一串**值"。先用两行确认它和"单个值"的区别。

````python
fruits = ["苹果", "香蕉", "梨"]
print(fruits[0])
print(len(fruits))
````

**预期输出**

```
苹果
3
```

- `fruits` 里装的是**三个**值，而不是一个；
- `fruits[0]` 取出第一个，`len(fruits)` 告诉我们一共几个。

本篇要做的事情：把这一串值的**编号规则**、**切片**、**增删改**，以及一个必须提前知道的坑，讲透。

---

## 1. 索引：第一个是 0

"第一个"的位置编号是 **0**，不是 1。

别记成"第零个"这种别扭的说法 —— 把方括号里的数字读成"**从开头往后偏移几位**"就顺了：偏移 0 位就是第一个。

````python
fruits = ["苹果", "香蕉", "梨"]
print(fruits[0])     # 偏移 0 位 = 第一个
print(fruits[1])     # 偏移 1 位 = 第二个
print(fruits[-1])    # 负数 = 从末尾往前数
print(fruits[-2])
````

**预期输出**

```
苹果
香蕉
梨
香蕉
```

- 正数从**开头**数，第一个是 `0`；
- 负数从**末尾**数，最后一个写 `-1`。

当你只关心"最后一个"而不确定列表有多长时，`fruits[-1]` 比"先 `len()` 再减 1"省事得多。

---

## 2. 切片：取一段，但**不含结尾位置**

写法是 `fruits[a:b]`，读作"**从 a 开始，到 b 之前为止**"。这是新手最常踩的一个坑。

````python
fruits = ["苹果", "香蕉", "梨", "葡萄"]
print(fruits[0:2])    # 编号 0、1 —— 不含 2
print(fruits[1:3])    # 编号 1、2 —— 不含 3
print(fruits[0:])     # 省略结尾 = 一直取到末尾
print(fruits[-2:])    # 从倒数第二个一直取到末尾
````

**预期输出**

```
['苹果', '香蕉']
['香蕉', '梨']
['苹果', '香蕉', '梨', '葡萄']
['梨', '葡萄']
```

记住一句就够：**切片的结尾是不包含的**。`fruits[0:2]` 取两个元素，不是三个。

想取到最后一个，要么把结尾写全（`fruits[0:4]`），要么干脆省略结尾（`fruits[0:]`）。

## 3. 长度与"在不在"

`len(a)` 得到元素个数；`x in a` 判断某个值在不在里面 —— 结果永远是 `True` 或 `False`。

````python
fruits = ["苹果", "香蕉", "梨"]
print(len(fruits))
print("香蕉" in fruits)
print("葡萄" in fruits)
````

**预期输出**

```
3
True
False
```

两件事都重要：

- `len()` 是**防止越界**的依据 —— 想在列表里取第 5 个，先确认 `len()` 至少是 5；
- `in` 让"在不在"变成一个可以直接用的真假值。

> **成员运算 `in` 归本篇**（以及《字典》篇）；《运算符》篇只讲算术、比较、逻辑三组，不含 `in`。

## 4. 增 / 删 / 改：列表是"活的"

列表创建之后还能**原地**增删改。五个动作，各管一件事。

````python
todo = ["买菜", "洗衣"]
todo.append("写代码")       # 末尾加一项
print(todo)

todo.insert(0, "晨跑")      # 插到指定位置
print(todo)

todo.remove("洗衣")         # 按「值」删
print(todo)

last = todo.pop()          # 按「位置」删，并把删掉的那一项返回
print("取出的：", last)
print(todo)

del todo[0]                # 按「索引」删
print(todo)
````

**预期输出**

```
['买菜', '洗衣', '写代码']
['晨跑', '买菜', '洗衣', '写代码']
['晨跑', '买菜', '写代码']
取出的： 写代码
['晨跑', '买菜']
['买菜']
```

| 写法 | 做什么 | 按什么定位 |
| --- | --- | --- |
| `append(值)` | 末尾加一项 | — |
| `insert(位置, 值)` | 插到指定位置 | 位置 |
| `remove(值)` | 删掉这个**值** | 值 |
| `pop(位置)` | 删掉这个位置的元素，并把它**返回** | 位置 |
| `del a[位置]` | 删掉这个位置的元素 | 位置 |

**一句关键提醒**：这些方法**改的是列表本身，不返回新列表**。所以千万别写 `todo = todo.append("X")` —— 那样 `todo` 会变成 `None`。这条规律会在下一节的 `sort()` 上再出现一次。

> `remove("洗衣")` 时如果列表里没有这个值，会抛 `ValueError`。

## 5. 排序：`sort()` 改原件，`sorted()` 出副本

两者排序结果看起来一样，但对**原列表**做的手脚完全不同。

````python
scores = [88, 95, 72]
result = scores.sort()              # 原地排序
print("sort() 的返回值：", result)
print("原列表被改成：", scores)

original = [88, 95, 72]
ranked = sorted(original, reverse=True)   # 返回一个新列表
print("sorted() 返回的新列表：", ranked)
print("原列表没被动：", original)
````

**预期输出**

```
sort() 的返回值： None
原列表被改成： [72, 88, 95]
sorted() 返回的新列表： [95, 88, 72]
原列表没被动： [88, 95, 72]
```

| 写法 | 改动谁 | 返回值 |
| --- | --- | --- |
| `a.sort()` | **原地改 a** | `None` |
| `sorted(a)` | a 不动 | 排好序的**新列表** |

`sort()` 返回 `None` 是刻意设计的，它在提醒你"排序已经发生在原件上了"。所以 `a = a.sort()` 会把 `a` 变成 `None` —— 这是高频事故。

**怎么选**：原顺序还要留着 → 用 `sorted()`；原顺序没用了 → 用 `sort(reverse=True)`，省下一份内存。

## 6. 可变性：`b = a` 只是多了个名字

上半场学过"变量是名字指向值"。现在把这个原理放到**列表**（可变对象）上 —— 会出现一个让人吃惊的现象。

````python
a = [1, 2, 3]
b = a                  # 不是复制！只是给同一个列表起了第二个名字
a.append(4)

print("b =", b)
print("a 和 b 是同一个列表吗？", a is b)

c = a.copy()           # 想要独立副本，得显式复制
a.append(5)
print("c =", c)
print("a =", a)
````

**预期输出**

```
b = [1, 2, 3, 4]
a 和 b 是同一个列表吗？ True
c = [1, 2, 3, 4]
a = [1, 2, 3, 4, 5]
```

**我明明没动 `b`，为什么 `b` 也变了？**

因为 `b = a` **没有复制出任何新列表** —— `a` 和 `b` 是**同一个列表**的两个名字。`a.append(4)` 改的是那个列表本身，所以从 `b` 看过去也是 4 个元素。

想要真正独立的一份，得**显式复制**：

- `a.copy()`
- `a[:]`（用切片复制整份）

复制出来的 `c` 就与 `a` 各走各的路了 —— 后面 `a.append(5)` 完全没影响到 `c`。

> `is` 判断"是不是同一个对象"，`==` 判断"内容是否相等"。这是两个不同的问题；本篇只需知道 `is` 能帮我们确认"同名同物"。
>
> 这条原理的完整讲解在本篇；《字典》篇只提一句"字典同样可变"，不重复展开。

## 7. 越界：`IndexError`

列表只有 3 个元素时，去取"第 100 个"会发生什么？先看正常的那一行。

````python
fruits = ["苹果", "香蕉", "梨"]
print(fruits[0])

# 把下面这一行行首的 # 去掉再运行，看看会发生什么：
# print(fruits[99])
````

**预期输出**（注释保留时）

```
苹果
```

把最后那一行的 `#` 去掉再运行，你会看到：

```
---------------------------------------------------------------------------
IndexError                                Traceback (most recent call last)
Cell In[8], line 5
----> 5 print(fruits[99])

IndexError: list index out of range
```

`IndexError: list index out of range` 的意思是：**编号超出范围了**。

这里要记住的是**结论**，不是读报错的技巧：列表只有 `len()` 个位置，编号从 `0` 到 `len() - 1`；超出就一定越界。

> 完整的 Traceback 逐行读法与排查动作，见补充章节《报错怎么读》。本篇只负责让你认出"越界"这一类问题。
