## 1.3 字符串操作

字符串是数据分析中**出现频率最高**的类型，因为原始数据里的分类字段、日期字段、文本字段都是字符串。

### 索引与切片

Python 的索引**从 0 开始**，负数表示从末尾倒数。

````python
s = "2024-01-15"

print(s[0])       # 第 1 个字符
print(s[-1])      # 最后 1 个字符
print(s[0:4])     # 第 0 到第 3 个字符
print(s[5:7])     # 第 5 到第 6 个字符
````

**预期输出**

```
2
5
2024
01
```

切片规则为 `s[起:止]`，**取值包含"起"、不包含"止"**。这是 Python 中非常重要的一个约定。

可以看到 `s[0:4]` 取到的是 `2024`——即下标 0、1、2、3 四个字符，下标 4 的 `-` 不取。请把 `s[5:7]` 改成 `s[5:8]` 再运行一次，观察结果怎么变。

### 常用字符串方法

````python
text = "  张三, 20, 男  "

print(text.strip())              # 去掉两端空白
print(text.strip().split(","))   # 按逗号拆分
print("ABC".lower())             # 转小写
print("abc".upper())             # 转大写
print("张三".replace("张", "李")) # 替换
print(len("Python"))             # 长度
````

**预期输出**

```
张三, 20, 男
['张三', ' 20', ' 男']
abc
ABC
李三
6
```

> **注意**：`split()` 的结果是一个**列表**（第 2 课内容），且拆分后每段可能带空格——你看第二行的输出里 `' 20'` 前面就有个空格。所以实际写法常写成 `[x.strip() for x in text.split(",")]` 的形式。

### f-string 格式化

这是当前**推荐**的字符串拼接方式，写法简洁、可读性好。

````python
name = "张三"
score = 92.5

print(f"{name} 的成绩是 {score} 分")
print(f"平均分保留两位小数：{score:.2f}")
````

**预期输出**

```
张三 的成绩是 92.5 分
平均分保留两位小数：92.50
```

**f-string 的三个好处**（对比 `"我的成绩是 " + str(score) + " 分"`）：

1. **不用手动 `str()`** —— `{score}` 里放数字也行，Python 自动处理。
2. **能直接格式化** —— `{score:.2f}` 保留两位小数，`{score:>8}` 右对齐。
3. **能塞表达式** —— `f"{score + 5}"`、`f"{len(name)}"` 都可以。

请把 `{score:.2f}` 改成 `{score:.0f}`（不要小数）和 `{score:.4f}`（四位小数），各运行一次看看区别。

## 1.4 运算符

### 算术运算符

| 运算符 | 含义 | 示例 | 结果 |
| --- | --- | --- | --- |
| `+` | 加 | `7 + 2` | `9` |
| `-` | 减 | `7 - 2` | `5` |
| `*` | 乘 | `7 * 2` | `14` |
| `/` | 除 | `7 / 2` | `3.5` |
| `//` | 整除（向下取整） | `7 // 2` | `3` |
| `%` | 取余 | `7 % 2` | `1` |
| `**` | 幂 | `2 ** 3` | `8` |

其中 `//` 和 `%` 需要特别理解，它们在数据分析中常用于**分组**和**取周期**。

````python
total = 100
groups = 7

print(f"每组 {total // groups} 个，还剩 {total % groups} 个")
````

**预期输出**

```
每组 14 个，还剩 2 个
```

**理解方式**：`100 // 7` 问的是"能装满几组"，`100 % 7` 问的是"装完还剩几个"。把 `total` 改成 `50` 再运行一次，验证你的理解。

> 顺带记住一个细节：`/` 的结果**永远是浮点数**。`4 / 2` 得到的是 `4.0` 而不是 `4`。如果后续做计数、做下标，需要整数就用 `//`。

### 比较运算符与逻辑运算符

````python
score = 85

print(score >= 60)                        # 是否及格
print(score >= 60 and score < 90)         # 是否在 60 到 90 之间
print(score < 60 or score > 100)          # 是否异常
print(not (score >= 60))                  # 是否不及格
````

**预期输出**

```
True
True
False
False
```

> **注意两点**：
> 1. 比较运算符的结果是布尔值 `True` / `False`，这正是第 2 课条件判断的基础。
> 2. `=` 是**赋值**，`==` 才是**判断相等**，两者不能混用。写成 `if score = 60` 会直接报语法错误。

## 1.5 输入输出与注释

### 输入

`input()` 会把用户输入的内容**一律当作字符串**返回。

````python
name = input("请输入姓名：")
score = input("请输入成绩：")

print(type(score))
````

**运行这一格时，下方会出现输入框**（Notebook 里输入框会出现在单元格下方）。输入"张三"回车、"88"回车。

若输入姓名为"张三"、成绩为"88"，预期输出：

```
请输入姓名：张三
请输入成绩：88
<class 'str'>
```

**看清最后一行**：`score` 明明是 `88`，但类型是 `str`。这就是为什么读取数字输入时必须手动转换。

````python
score = float(input("请输入成绩："))
print(f"成绩加了 5 分后是 {score + 5}")
````

这次用了 `float(input(...))` —— 先取输入，再转浮点数，最后参与运算。**这是最常见的两步走写法，必须练熟。**

如果这一步报错，多半是你输入了非数字字符，会看到：

```
ValueError: could not convert string to float: 'abc'
```

`ValueError` 的意思是"类型对了（都是字符串），但值本身没法转换"。**注意区分 `TypeError` 和 `ValueError`**，这是报错排查的关键一步。

### 输出格式化的常用写法

````python
print("姓名", "成绩", sep=" | ")
print("第一行\n第二行")
print("结果", end="")
print("在同一行")
````

**预期输出**

```
姓名 | 成绩
第一行
第二行
结果在同一行
```

- `sep=` 决定多个值之间用什么分隔（默认是一个空格）
- `\n` 是换行符
- `end=` 决定结尾用什么（默认是 `\n`，改成空字符串就不会换行了）

### 注释

注释不参与执行，用于解释代码意图。

````python
# 这是单行注释

"""
这是多行注释
常用于说明函数用途
"""

total = 0  # 也可以写在代码行末尾
print("total =", total)
````

**养成一个习惯**：每写完一段能跑通的逻辑，就补一句注释说明它在干什么。等到第 3 课开始写函数、第 4 课写分析流程时，这个习惯能省下大量回头读代码的时间。
